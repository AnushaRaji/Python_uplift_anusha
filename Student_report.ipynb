```python
# 1. Data Structure:
#    Create a dictionary to store students with this structure:
#    {
#        "S001": {
#            "name": "Alice",
#            "grades": [85, 90, 78, 92],  # list of test scores
#            "attendance": 0.90,  # percentage as decimal (0-1)
#            "projects": {"proj1": 95, "proj2": 88}
#        }
#    }

students = {
    "S001": {
        "name": "Alice",
        "grades": [85, 90, 78, 92],
        "attendance": 0.90,
        "projects": {"proj1": 95, "proj2": 88}
    },
    "S002": {
        "name": "Bob",
        "grades": [80, 85, 88, 90],
        "attendance": 0.95,
        "projects": {"proj1": 80, "proj2": 85}
    },
    "S003": {
        "name": "Charlie",
        "grades": [70, 75, 80, 85],
        "attendance": 0.85,
        "projects": {"proj1": 70, "proj2": 75}
    }   

}
# Functions to Write:
#    - add_student(students, student_id, name)
def add_student(students, student_id, name):
    if student_id not in students:
        students[student_id] = {
            "name": name,
            "grades": [],
            "attendance": 0.0,
            "projects": {}
        }

add_student(students, "S004", "David")
add_student(students, "S005", "Eve")
add_student(students, "S006", "Frank")

#print("\n students", students)

#    - add_grade(students, student_id, grade)
def add_grade(students, student_id, grade):
    if student_id in students:
        students[student_id]["grades"].extend(grade)
add_grade(students, "S004", [88, 91, 84, 89, 93])
add_grade(students, "S005", [92, 88, 90, 87, 95]) 
add_grade(students, "S006", [78, 82, 75, 80, 79])
#print("\n students", students)

#    - update_attendance(students, student_id, attendance)
def update_attendance(students, student_id, attendance):
    if student_id in students:
        students[student_id]["attendance"] = attendance
update_attendance(students, "S004", 0.92)
update_attendance(students, "S005", 0.88)
update_attendance(students, "S006", 0.91)       
#print("\n students", students)
#    - add_project_score(students, student_id, project_name, score)
def add_project_score(students, student_id, project_name, score):
    if student_id in students:
        students[student_id]["projects"][project_name] = score
add_project_score(students, "S004", "proj1", 90)
add_project_score(students, "S004", "proj2", 85)
add_project_score(students, "S005", "proj1", 88)
add_project_score(students, "S005", "proj2", 92)
add_project_score(students, "S006", "proj1", 80)
add_project_score(students, "S006", "proj2", 78)
print("\n students", students)

def calculate_test_average(students, student_id):
    if student_id in students:
        grades = students[student_id]["grades"]
        if grades:
            avg = sum(grades) / len(grades)
            return round(avg, 2)
    return 0
# print(" Average grade for S001:", calculate_test_average(students, "S001"))
# print(" Average grade for S002:", calculate_test_average(students, "S002"))
# print(" Average grade for S003:", calculate_test_average(students, "S003"))
# print("\n Average grade for S004:", calculate_test_average(students, "S004"))
# print(" Average grade for S005:", calculate_test_average(students, "S005"))
# print(" Average grade for S006:", calculate_test_average(students, "S006"))

def calculate_project_average(students, student_id):
    projects = students[student_id]["projects"]
    if not projects:
        return 0.0
    scores = projects.values()
    avg1 = sum(scores) / len(scores)
    return round(avg1, 2)
# print(" Project average for S001:", calculate_project_average(students, "S001"))
# print(" Project average for S002:", calculate_project_average(students, "S002"))
# print(" Project average for S003:", calculate_project_average(students, "S003"))
# print("\n Project average for S004:", calculate_project_average(students, "S004"))
# print(" Project average for S005:", calculate_project_average(students, "S005"))
# print(" Project average for S006:", calculate_project_average(students, "S006"))


# calculate_final_score(students, student_id)
#      Formula: Tests 60% + Projects 25% + Attendance 15%
#      (Attendance score = attendance * 100)

def calculate_final_score(students, student_id):
    test_avg = calculate_test_average(students, student_id)
    project_avg = calculate_project_average(students, student_id)
    attendance = students[student_id]["attendance"] * 100
    final_score = (test_avg * 0.6) + (project_avg * 0.25) + (attendance * 0.15)
    return round(final_score, 2)
print(" Final score for S001:", calculate_final_score(students, "S001"))
print(" Final score for S002:", calculate_final_score(students, "S002"))
print(" Final score for S003:", calculate_final_score(students, "S003"))
print("\n Final score for S004:", calculate_final_score(students, "S004"))
print(" Final score for S005:", calculate_final_score(students, "S005"))
print(" Final score for S006:", calculate_final_score(students, "S006"))

# get_letter_grade(score)
#      A: 90+, B: 80-89, C: 70-79, D: 60-69, F: <60
def get_letter_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'
print(" Letter grade for S001:", get_letter_grade(calculate_final_score(students, "S001")))
print(" Letter grade for S002:", get_letter_grade(calculate_final_score(students, "S002")))
print(" Letter grade for S003:", get_letter_grade(calculate_final_score(students, "S003")))
print("\n Letter grade for S004:", get_letter_grade(calculate_final_score(students, "S004")))
print(" Letter grade for S005:", get_letter_grade(calculate_final_score(students, "S005")))
print(" Letter grade for S006:", get_letter_grade(calculate_final_score(students, "S006")))

#- Class average
def class_average(students):
    total = 0
    count = 0
    for student_id in students:
        total += calculate_final_score(students, student_id)
        count += 1
    if count == 0:
        return 0.0
    avg = total / count
    return round(avg, 2)
print("\n Class average score:", class_average(students))

#Find top 3 students
def top_students(students, n=3):
    scores = []
    for student_id in students:
        final_score = calculate_final_score(students, student_id)
        scores.append((final_score, student_id))
    scores.sort(reverse=True)
    return scores[:n]

# 4. Generate Report:
#    Calculate and display:
#    - Each student's test average, project average, and final score
#    - Letter grade for each student
#    - Class average
#    - Top 3 performing students
#    - Students at risk (final score < 60)
top_3 = top_students(students)
print("\n Top 3 students:")
for score, student_id in top_3:
    print(f" Student ID: {student_id}, Final Score: {score}, Name: {students[student_id]['name']}")
print("\n Students at risk (final score < 60):")
for student_id in students:
    final_score = calculate_final_score(students, student_id)
    if final_score < 60:
        print(f" Student ID: {student_id}, Final Score: {final_score}, Name: {students[student_id]['name']}")
    else:
        print(" No students at risk")



# 5. Save Results:
#    - Save the report to a text file with formatted output

# EXPECTED OUTPUT FORMAT:
# === STUDENT GRADE REPORT ===

# Student: Alice (ID: S001)
#   Test Average: 86.25
#   Project Average: 91.50
#   Attendance: 90.0%
#   Final Score: 87.43
#   Letter Grade: B

# Class Statistics:
#   Class Average: 82.15
#   Top 3: Alice (87.43), Bob (85.20), Charlie (83.10)
#   At Risk: David (58.50)

# Report saved to: student_report.txt

with open("student_report.txt", "w") as f:
    f.write("=== STUDENT GRADE REPORT ===\n\n")
    for student_id in students:
        name = students[student_id]["name"]
        test_avg = calculate_test_average(students, student_id)
        project_avg = calculate_project_average(students, student_id)
        attendance = students[student_id]["attendance"] * 100
        final_score = calculate_final_score(students, student_id)
        letter_grade = get_letter_grade(final_score)
        f.write(f"Student: {name} (ID: {student_id})\n")
        f.write(f"  Test Average: {test_avg}\n")
        f.write(f"  Project Average: {project_avg}\n")
        f.write(f"  Attendance: {attendance:.1f}%\n")
        f.write(f"  Final Score: {final_score}\n")
        f.write(f"  Letter Grade: {letter_grade}\n\n")
    
    f.write("Class Statistics:\n")
    f.write(f"  Class Average: {class_average(students)}\n")
    f.write("  Top 3: ")
    top_3_list = []
    for score, student_id in top_3:
        top_3_list.append(f"{students[student_id]['name']} ({score})")
    f.write(", ".join(top_3_list) + "\n")
    
    at_risk_list = []
    for student_id in students:
        final_score = calculate_final_score(students, student_id)
        if final_score < 60:
            at_risk_list.append(f"{students[student_id]['name']} ({final_score})")
    f.write("  At Risk: ")
    if at_risk_list:
        f.write(", ".join(at_risk_list) + "\n")
    else:
        f.write("None\n")

        








```
