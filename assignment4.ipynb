```python
# 1. A 1D array of numbers from 0 to 9
array1 = np.array([range(10)])
print(f'1D array : {array1}')

# 2. A 2D array of shape (3, 4) filled with zeros
array2 = np.zeros((3, 4))
print(f'3x4 array of Zero values : \n{array2}')

# 3. A 3x3 identity matrix
array3 = np.eye(3)
print(f'3x3 Identity matrix : \n{array3}')

# 4. A 1D array of 10 equally spaced values between 1 and 100
array4 = np.linspace(1, 100, 10)
print(f'1D array of 10 equally spaced values between 1 and 100 : \n{array4}')

# 5. A 3x3 array with random integers between 10 and 50
array5 = np.random.randint(10, 51, size=(3, 3))
print(f'3x3 array with random integers between 10 and 50 : \n{array5}')


```

```python

# Create a 3D array with shape (2, 3, 4) using random numbers
#array6 = np.random.randint(10, 51, size=(2, 3, 4))
array6 = np.random.randn(2, 3, 4)

print(f'3D array with shape (2, 3, 4) : \n{array6}')

# Find and print:
# - Number of dimensions
# - Shape of the array
# - Total number of elements
# - Data type of elements
# - Size in bytes of each element
# - Total memory consumed by the array
print(f'Number of dimensions: {array6.ndim}')
print(f'Shape of the array: {array6.shape}')
print(f'Total number of elements: {array6.size}')
print(f'Data type of elements: {array6.dtype}')
print(f'Size in bytes of each element: {array6.itemsize}')
print(f'Total memory consumed by the array: {array6.nbytes} bytes')
```

```python
# Exercise 3: 
# Create a 1D array of numbers from 1 to 20
# Perform the following operations:
# - Access the first, last, and middle elements
# - Slice the array to get elements from index 5 to 15
# - Get all even numbers using slicing
# - Reverse the array using slicing

array7 = np.array(range(1, 21))
print(f'Original array: {array7}')
print(f'First element: {array7[0]}')
print(f'Last element: {array7[-1]}')
print(f'Middle element: {array7[len(array7)//2]}')
print(f'Elements from index 5 to 15: {array7[5:16]}')
print(f'All even numbers: {array7[array7 % 2 == 0]}')
print(f'Reversed array: {array7[::-1]}')
```

```python
# Exercise 4: 
# Create a 5x5 array with numbers from 1 to 25
# Perform the following tasks:
# - Access the element at row 2, column 3
# - Extract the third row and second column
# - Slice a 3x3 sub-matrix from the center
# - Replace all values in the last row with 0
# - Change the first column to all -1
array8 = np.array(range(1, 26)).reshape(5, 5)
print(f'Original 5x5 array: \n{array8}')
print(f'Element at row 2, column 3: {array8[1, 2]}')
print(f'Third row: {array8[2, :]}')
print(f'Second column: {array8[:, 1]}')
print(f'3x3 sub-matrix from the center: \n{array8[1:4, 1:4]}')
array8[-1, :] = 0
print(f'Array after replacing last row with 0: \n{array8}')
array8[:, 0] = -1
print(f'Array after changing first column to -1: \n{array8}')

  
```

```python
# Exercise 5: 
# Create a 1D array of 15 random integers between 1 and 50
# Perform the following:
# - Select elements at indices [1, 3, 5, 7]
# - Find all elements greater than 25
# - Replace all values less than 10 with 10
# - Count how many values are between 20 and 40 (inclusive)
# - Extract only the odd numbers from the array

array9 = np.random.randint(1, 51, size=15)
print(f'Original array: {array9}')
print(f'Elements at indices [1, 3, 5, 7]: {array9[[1, 3, 5, 7]]}')
print(f'Elements greater than 25: {array9[array9 > 25]}')
array9[array9 < 10] = 10
print(f'Array after replacing values less than 10 with 10: {array9}')
count_20_40 = np.sum((array9 >= 20) & (array9 <= 40))
print(f'Count of values between 20 and 40 (inclusive): {count_20_40}')
# - Extract only the odd numbers from the array
odd_numbers = array9[array9 % 2 != 0]
print(f'Odd numbers in the array: {odd_numbers}')
```
