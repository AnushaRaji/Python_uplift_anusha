```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('whitegrid')
```

```python
np.random.seed(42)
x=np.random.randn(100)
y=2 * x + np.random.randn(100)
plt.scatter(x,y, alpha=0.7, color='blue', s=40)
plt.title('Scatter Plot of y vs x')
plt.xlabel('x values')
plt.ylabel('y values')
plt.show()
```

```python

x=[1,2,3,4,5]
y=[20,30,25,35,30]
plt.barh(x,y, alpha=0.7, color='blue')
plt.title('Bar Plot of y vs x')
plt.xlabel('x values')
plt.ylabel('y values')
plt.show()
```
