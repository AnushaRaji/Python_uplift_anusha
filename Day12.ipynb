```python
# TASK 2.1: Statistical Analysis
# For each batsman, use NumPy to calculate:

# 1. Convert runs_scored list to NumPy array
# 2. Calculate:
#    - Mean runs per match
#    - Standard deviation of runs
#    - Median runs
#    - Highest and lowest score
#    - Consistency score = (mean / std) - higher is more consistent

# 3. Create a 4×14 matrix (4 batsmen × 14 matches) with all runs
# 4. Calculate:
#    - Total runs by each player (sum along axis)
#    - Average runs per match across all batsmen
#    - Match-wise team total (sum along axis)

import numpy as np
batsman_ids = [pid for pid, pdata in players.items() if pdata["role"] == "Batsman"]
batsman_runs = np.array([players[pid]["runs_scored"] for pid in
                        batsman_ids])
for i, pid in enumerate(batsman_ids):
    runs = batsman_runs[i]
    mean_runs = np.mean(runs)
    std_runs = np.std(runs)
    median_runs = np.median(runs)
    highest_score = np.max(runs)
    lowest_score = np.min(runs)
    consistency_score = mean_runs / std_runs if std_runs != 0 else float('inf')
    
    print(f"Statistics for {players[pid]['name']}:")
    print(f" Mean Runs: {mean_runs:.2f}")
    print(f" Standard Deviation: {std_runs:.2f}")
    print(f" Median Runs: {median_runs}")
    print(f" Highest Score: {highest_score}")
    print(f" Lowest Score: {lowest_score}")
    print(f" Consistency Score: {consistency_score:.2f}\n")
    

```
